# SIMPLIFICAR E RENOMEAR 
    Nesta etapa vamos simplificar a tabela para que se mantenha apenas as informações que vamos utilizar para explorar a base e renomeá-las para a praticidade da codificação do projeto. 

```python
tabela = tabela [
    ["player_name", "age", "position", "goals_2025_26", "assists_2025_26", "market_value_million", "club", "nationality"
     ]
]
```

```python
tabela = tabela.rename(columns={
    "player_name": "jogador",
    "age": "idade",
    "position": "posicao",
    "goals_2025_26": "gols",
    "assists_2025_26": "assistencias",
    "market_value_million": "valor_mercado",
    "Club": "clube",
    "nationality": "nacionalidade"
})

tabela
```

# JOGADOR QUE MAIS MARCOU GOLS NESTA TEMPORADA?

```python
tabela.loc[tabela["gols"].idxmax()]
```

# RANKING DOS 10 JOGADORES

```python
top_goleadores = tabela.sort_values(
    by="gols",
    ascending=False
)

top_goleadores.head(10)
```

# TOP 10 JOGADORES POR VALOR DE MERCADO:

Nesta análise, os jogadores foram classificados de acordo com o valor de mercado, identificando os dez atletas mais valorizados da base de dados. Essa análise permite observar quais jogadores possuem maior valorização financeira na temporada analisada.

```python
top_valor_mercado = tabela.sort_values(
    by="valor_mercado",
    ascending=False 
)
top_valor_mercado
```
